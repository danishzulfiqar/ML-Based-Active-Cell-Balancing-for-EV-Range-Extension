```python
# Getting the list of all the files in the directory
files = os.listdir('../../DataSets/Complete_Data')

# files filter only mat files
files = [file for file in files if '.mat' in file]

print(f"Total files: {len(files)}")
```

```python
print(len(files))
```

```python
# Initialize an empty DataFrame
df = pd.DataFrame(columns=['w1', 'w2', 'w3', 'wPL', 'DC' , 'range'])
```

```python
# List to collect rows for the DataFrame
rows = []

# Iterating over all the files and adding the values to the rows list
for file in files:
    file_path = os.path.join('../../DataSets/Complete_Data', file)
    try:
        # Check if the file size is greater than zero
        if os.path.getsize(file_path) > 0:
            data = loadmat(file_path)
            # Check if all required keys are present in the file
            if 'w1' in data and 'w2' in data and 'w3' in data and 'w_PL' in data and 'range' in data and 'DC' in data:
                # Append a new row to the list
                rows.append({
                    'w1': data['w1'][0][0], 
                    'w2': data['w2'][0][0], 
                    'w3': data['w3'][0][0], 
                    'wPL': data['w_PL'][0][0], 
                    # adding DC which is a string 
                    'DC': data['DC'][0],
                    'range': data['range'][-1][0],
                })
            else:
                print(f"Skipping file with missing data: {file}")
        else:
            print(f"Skipping empty file: {file}")
    except Exception as e:
        print(f"Error reading {file}: {e}")
```

```python
# Creating the DataFrame from the collected rows
df = pd.DataFrame(rows)
```

```python
# Display the final DataFrame shape and check for duplicates
print(f"DataFrame shape: {df.shape}")
print(f"Number of unique rows: {df.drop_duplicates().shape[0]}")
```

```python
# printing duplicate rows

print(df[df.duplicated()])
```

```python
# DC
# FUDSfast_extended    591
# US06_extended        429
# FUDS_extended        402
# US06City_extended    401
# NEDC_extended        400
# FHDS_extended        400
# US06HW_extended      400

# making all dc datapoints to 400 each
# removing random 191 datapoints from FUDSfast_extended
# removing random 29 datapoints from US06_extended
# removing random 2 datapoints from US06City_extended
# removing random 1 datapoints from NEDC_extended


df = df.drop(df[(df['DC'] == 'FUDSfast_extended')].sample(191).index)
df = df.drop(df[(df['DC'] == 'US06_extended')].sample(29).index)
df = df.drop(df[(df['DC'] == 'FUDS_extended')].sample(2).index)
df = df.drop(df[(df['DC'] == 'US06City_extended')].sample(1).index)

print(df['DC'].value_counts())
```

```python
# max value of range for each DC

data_unique = df.drop_duplicates().reset_index(drop=True)
max_range_rows = data_unique.loc[data_unique.groupby('DC')['range'].idxmax()]
print(max_range_rows)
```

```python
# Saving the cleaned DataFrame to a CSV file
max_range_rows.to_csv('./Python/max_range_in_dataset.csv', index=False)
```
