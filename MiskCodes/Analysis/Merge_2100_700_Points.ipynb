```python
# importing libraries

from scipy.io import loadmat
import os
import numpy as np
import pandas as pd
```

```python
# reading two csv files
data_700 = pd.read_csv("./DataSet_Encoded_DriveCycles.csv")
data_2100 = pd.read_csv("./2093_DataPoints_With_DC.csv")
```

```python
# Printing the shape of the data
print(data_700.shape)
print(data_2100.shape)
```

```python
# printing the columns dataframes for both the data
print(data_700.columns)
print(data_2100.columns)
```

```python
pd.DataFrame(data_700).head()
```

```python
pd.DataFrame(data_2100).head()
```

```python
# finding duplicate rows in the data
duplicateRowsDF = data_700[data_700.duplicated()]
print(duplicateRowsDF)
```

```python
duplicateRowsDF = data_2100[data_2100.duplicated()]
print(duplicateRowsDF)
```

```python
# making new dataframe with both the data
data = pd.concat([data_700, data_2100], axis=0)

# printing the shape of the new dataframe
print(data.shape)
```

```python
pd.DataFrame(data).head()
```

```python
# finding duplicate rows in the data
duplicateRowsDF = data[data.duplicated()]
print(duplicateRowsDF)
```

```python
# seeing dupolicate rows DC counts
duplicateRowsDF['DC'].value_counts()
```

```python
# retaining unique rows in the data
data = data.drop_duplicates()
```

```python
# printing the shape of the data
print(data.shape)
```

```python
data['DC'].value_counts()
```

```python
data_unique = data.drop_duplicates().reset_index(drop=True)
max_range_rows = data_unique.loc[data_unique.groupby('DC')['range'].idxmax()]
print(max_range_rows)
```
