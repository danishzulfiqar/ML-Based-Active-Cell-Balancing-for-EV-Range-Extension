```python
import tensorflow as tf
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
# importing dataset from csv file

data = pd.read_csv('./Optimised_Bayesian_With_Max_Range.csv')
```

```python
pd.DataFrame(data).head()
```

```python
# Importing DC_Encoded CSV file

encoded_data = pd.read_csv('./Encoded_DC.csv')
```

```python
pd.DataFrame(encoded_data).head()
```

```python
# Adding DC_Name column to the main dataset from encoded dataset ie numeric value corresponding DC_Name

# DC,Encoded_DC
# NEDC_extended,3
# FHDS_extended,0
# US06HW_extended,5
# FUDSfast_extended,2
# US06City_extended,4
# US06_extended,6
# FUDS_extended,1


# Comparing the Encoded_DC values with the DC values and adding the DC_Name column to the main dataset


# Create a dictionary from the encoded_data dataframe
dc_dict = dict(zip(encoded_data['Encoded_DC'], encoded_data['DC']))

# Map the Encoded_DC values to DC_Name using the dictionary
data['DC_Name'] = data['DC'].map(dc_dict)

# Display the updated dataframe
data.head()
```

```python
# re arranging the columns

# w1	w2	w3	wPL	DC DC_Name	Optimal_Range	Max_Range_In_Data	Max_Predicted_Range	

new_data = data[['w1', 'w2', 'w3', 'wPL', 'DC', 'DC_Name', 'Optimal_Range', 'Max_Range_In_Data', 'Max_Predicted_Range']]
```

```python
pd.DataFrame(new_data)
```

```python
# converting wPL to string instead of numerical values

new_data['wPL'] = new_data['wPL'].astype(str)
```

```python
# Saving the new data to a csv file

new_data.to_csv('Final_DNN_Optamised_Outputs_Summary.csv', index=False)
```
