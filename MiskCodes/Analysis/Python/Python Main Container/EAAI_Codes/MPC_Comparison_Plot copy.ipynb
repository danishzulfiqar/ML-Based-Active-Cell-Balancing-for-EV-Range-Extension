```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Import dataset
df = pd.read_csv('./MPC_Comparison.csv')

# Extract unique DC_Name values
DC = df['DC_Name'].unique()

# Prepare data for plotting
w1_values = []
w2_values = []
w3_values = []
mpc_values = []

for dc in DC:
    data = df[df['DC_Name'] == dc]
    w1_values.append(data[data['w1'] == 1]['Range'].values[0])
    w2_values.append(data[data['w2'] == 1]['Range'].values[0])
    w3_values.append(data[data['w3'] == 1]['Range'].values[0])
    mpc_values.append(data['MPC_Range_from_Optamised_Weights'].values[0])

# Bar width
bar_width = 0.2
x_labels = DC

# Set position of bars on X axis
r1 = np.arange(len(DC))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]
r4 = [x + bar_width for x in r3]


# Define colors for bars
colors = ["#FF4C4C", "#4CAF50", "#3B82F6", "#F5C518"]

# Plot grouped bar chart
plt.figure(figsize=(16,9))
bars1 = plt.bar(r1, w1_values, color=colors[0], width=bar_width, edgecolor='black', label='L1 (w1)')
bars2 = plt.bar(r2, w2_values, color=colors[1], width=bar_width, edgecolor='black', label='L2 (w2)')
bars3 = plt.bar(r3, w3_values, color=colors[2], width=bar_width, edgecolor='black', label='L3 (w3)')
bars4 = plt.bar(r4, mpc_values, color=colors[3], width=bar_width, edgecolor='black', label="L' (Optimized Weights)")

# Add values on top of bars
for bars in [bars1, bars2, bars3, bars4]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height + 3, f'{height:.1f}', ha='center', va='bottom', fontsize=9, fontweight='normal', rotation=75)

# Add x labels
plt.xticks([r + bar_width * 1.5 for r in range(len(DC))], x_labels, rotation=0)

# Labels and legend
plt.xlabel('Drive Cycles (DC_Name)', fontweight='bold')
plt.ylabel('Range', fontweight='bold')
plt.legend()
plt.title("Comparison of Cost functions corresponding ranges for different drive cycles")


# Remove top and right borders
sns.despine()

plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set Seaborn style
sns.set_style("whitegrid")

# Import dataset
df = pd.read_csv('./MPC_Comparison.csv')

# Extract unique DC_Name values
DC = df['DC_Name'].unique()

# Prepare data for plotting
w1_values = []
w2_values = []
w3_values = []
mpc_values = []

for dc in DC:
    data = df[df['DC_Name'] == dc]
    w1_values.append(data[data['w1'] == 1]['Range'].values[0])
    w2_values.append(data[data['w2'] == 1]['Range'].values[0])
    w3_values.append(data[data['w3'] == 1]['Range'].values[0])
    mpc_values.append(data['MPC_Range_from_Optamised_Weights'].values[0])

# Define colors for bars
colors = ["#FF4C4C", "#4CAF50", "#3B82F6", "#F5C518"]

# Bar width
bar_width = 0.2
x_labels = DC

# Set position of bars on X axis
r1 = np.arange(len(DC))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]
r4 = [x + bar_width for x in r3]

# Plot grouped bar chart
plt.figure(figsize=(14, 10))
bars1 = plt.bar(r1, w1_values, color=colors[0], width=bar_width, edgecolor='black', label='L1 (w1)')
bars2 = plt.bar(r2, w2_values, color=colors[1], width=bar_width, edgecolor='black', label='L2 (w2)')
bars3 = plt.bar(r3, w3_values, color=colors[2], width=bar_width, edgecolor='black', label='L3 (w3)')
bars4 = plt.bar(r4, mpc_values, color=colors[3], width=bar_width, edgecolor='black', label="L' (Optimized Weights)")

# Add values on top of bars
for bars in [bars1, bars2, bars3, bars4]:
    for bar in bars:
        height = bar.get_height()
        plt.text(bar.get_x() + bar.get_width()/2, height + 5, f'{height:.1f}', ha='center', va='bottom', fontsize=9, fontweight='normal', rotation=90)

# Add x labels
plt.xticks([r + bar_width * 1.5 for r in range(len(DC))], x_labels, rotation=20, fontsize=12, fontweight='bold')

# Labels and legend
plt.xlabel('Drive Cycles (DC_Name)', fontweight='bold', fontsize=13)
plt.ylabel('Range', fontweight='bold', fontsize=13)
plt.title("Comparison of Cost Functions Corresponding Ranges for Different Drive Cycles", fontsize=14, fontweight='bold')

# Add legend with shadow
plt.legend( frameon=True, shadow=True)

# Add a grid for readability
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Remove top and right borders
sns.despine()

# Show plot
plt.show()
```
