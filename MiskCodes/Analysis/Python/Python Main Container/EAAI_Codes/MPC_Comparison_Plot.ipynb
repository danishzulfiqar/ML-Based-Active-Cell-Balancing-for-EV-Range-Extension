```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Import dataset
df = pd.read_csv('./MPC_Comparison.csv')
```

```python
# Extract unique DC_Name values
DC = df['DC_Name'].unique()

# Prepare data for plotting
w1_values = []
w2_values = []
w3_values = []

for dc in DC:
    data = df[df['DC_Name'] == dc]
    w1_values.append(data[data['w1'] == 1]['Range'].values[0])
    w2_values.append(data[data['w2'] == 1]['Range'].values[0])
    w3_values.append(data[data['w3'] == 1]['Range'].values[0])

# Bar width
bar_width = 0.2
x_labels = DC

# changing DC_Name to DC_1, DC_2, DC_3, DC_4 etc
for i in range(len(DC)):
    # print real and fake DC_Name
    print(DC[i] ," -----> ", 'DC ' + str(i + 1))
    DC[i] = 'DC ' + str(i + 1)

# Set position of bars on X axis
r1 = np.arange(len(DC))
r2 = [x + bar_width for x in r1]
r3 = [x + bar_width for x in r2]
```

```python
# Define colors for bars
colors = ["#FF4C4C", "#4CAF50", "#3B82F6"]

# Plot grouped bar chart
plt.figure(figsize=(16,9))
bars1 = plt.bar(r1, w1_values, color=colors[0], width=bar_width, edgecolor='black', label='L1')
bars2 = plt.bar(r2, w2_values, color=colors[1], width=bar_width, edgecolor='black', label='L2')
bars3 = plt.bar(r3, w3_values, color=colors[2], width=bar_width, edgecolor='black', label='L3')


# Add x labels
plt.xticks([r + bar_width * 1 for r in range(len(DC))], x_labels, fontsize=20)

# Labels and legend
plt.xlabel('Drive Cycles', fontsize=20, labelpad=20)
plt.ylabel('Range', fontsize=20, labelpad=20)

# Add title
# plt.title('Range of EVs with Different Drive Cycles', fontsize=16, pad=20)

plt.legend(fontsize=20)

# Adding Y limits
plt.ylim(130, 230)

plt.yscale('linear')
plt.tick_params(axis='y', labelsize=20)

plt.grid(False)

# Remove top and right borders
sns.despine()

plt.show()
```
