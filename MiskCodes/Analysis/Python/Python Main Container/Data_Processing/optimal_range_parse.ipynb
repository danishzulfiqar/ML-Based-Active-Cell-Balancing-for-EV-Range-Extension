```python
# importing pandas, numpy and matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('./DataResults.csv')
```

```python
pd.DataFrame(df)
```

```python
drive_cycles = ['FHDS_extended', 'FUDS_extended', 'FUDSfast_extended', 'NEDC_extended', 'US06City_extended', 'US06HW_extended', 'US06_extended']

results = df.sort_values(by='DC_Name', key=lambda x: x.map({DC_Name: i for i, DC_Name in enumerate(drive_cycles)}))
```

```python
pd.DataFrame(results)
```

```python
# saving wPL as string, so does not show as e- notation in csv

df['wPL'] = df['wPL'].apply(lambda x: '%.10f' % x)
```

```python
# converting DC_name to string i.e. 'value'

df['DC_Name'] = df['DC_Name'].apply(lambda x: f'{x}')
```

```python
pd.DataFrame(df)
```

```python
df['DC_Name'] = df['DC_Name'].apply(lambda x: f"'{x}'")
pd.DataFrame(df)
```

```python
# saving to csv only the columns we want
# skipping the index column and last 2 columns

df.drop(df.columns[[ -1, -2]], axis=1).to_csv('DataResults_sorted.csv', index=False)
```
